```python
import numpy as np
import matplotlib.pyplot as plt


def newton(f, df, x0, tol=1e-12, max_iter=50):
    """
    Newton's method with iteration history.

    Parameters
    ----------
    f : callable
        Function whose root is sought.
    df : callable
        Derivative of f.
    x0 : float
        Initial guess.
    tol : float
        Stopping tolerance based on |f(x_n)|.
    max_iter : int
        Maximum allowed iterations.

    Returns
    -------
    x : float
        Final approximation.
    n_iter : int
        Number of iterations used.
    history : list[float]
        Sequence of iterates.
    residuals : list[float]
        Sequence of |f(x_n)|.
    """
    x = float(x0)
    history = [x]
    residuals = [abs(f(x))]

    for n in range(1, max_iter + 1):
        dfx = df(x)
        if dfx == 0:
            raise ZeroDivisionError(f"Derivative vanished at iteration {n}, x = {x}")

        x = x - f(x) / dfx
        history.append(x)
        residuals.append(abs(f(x)))

        if abs(f(x)) < tol:
            return x, n, history, residuals

    return x, max_iter, history, residuals
```

```python
f = lambda x: x**3 - x - 2
df = lambda x: 3*x**2 - 1

root_n, iters_n, hist_n, res_n = newton(f, df, x0=1.5)

print("Newton root:", root_n)
print("Iterations:", iters_n)
print("Final residual:", res_n[-1])
```

```python
def bisection(f, a, b, tol=1e-12, max_iter=100):
    """
    Bisection method with iteration history.
    """
    fa, fb = f(a), f(b)
    if fa * fb >= 0:
        raise ValueError("Function must change sign on interval [a,b].")

    history = []
    residuals = []

    for n in range(1, max_iter + 1):
        c = 0.5 * (a + b)
        fc = f(c)

        history.append(c)
        residuals.append(abs(fc))

        if abs(fc) < tol or 0.5 * (b - a) < tol:
            return c, n, history, residuals

        if fa * fc < 0:
            b, fb = c, fc
        else:
            a, fa = c, fc

    return c, max_iter, history, residuals
```

```python
root_b, iters_b, hist_b, res_b = bisection(f, 1, 2)

print("Bisection root:", root_b)
print("Iterations:", iters_b)
print("Final residual:", res_b[-1])
```

```python
plt.figure(figsize=(8, 5))
plt.semilogy(range(1, len(res_b) + 1), res_b, marker='o', label='Bisection')
plt.semilogy(range(0, len(res_n)), res_n, marker='s', label='Newton')
plt.xlabel("Iteration")
plt.ylabel("Residual |f(x_n)|")
plt.title("Convergence Comparison: Bisection vs Newton")
plt.legend()
plt.grid(True)
plt.show()
```

```python
alpha = root_n
errors_n = [abs(x - alpha) for x in hist_n]

print(errors_n)
```

```python
orders = []
for n in range(1, len(errors_n) - 1):
    e_nm1 = errors_n[n - 1]
    e_n = errors_n[n]
    e_np1 = errors_n[n + 1]

    if e_nm1 > 0 and e_n > 0 and e_np1 > 0:
        p_est = np.log(e_np1 / e_n) / np.log(e_n / e_nm1)
        orders.append(p_est)

print("Estimated orders for Newton:")
print(orders)
```

```python
alpha_b = root_b
errors_b = [abs(x - alpha_b) for x in hist_b]

ratios_b = []
for n in range(len(errors_b) - 1):
    if errors_b[n] > 0:
        ratios_b.append(errors_b[n + 1] / errors_b[n])

print("Bisection error ratios:")
print(ratios_b[-10:])
```

Newton vs Bisection: Bisection method converges globally on any interval with f(a)f(b) < 0 but linear. Newton quadratically but heavily dependent on initial guess and non-zero derivative
